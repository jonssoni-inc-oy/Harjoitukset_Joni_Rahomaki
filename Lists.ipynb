```python
from random import randint
thrownDiceNum=[2,5,6,1,3,]
for heitto in range(5):
    thrownDiceNum[heitto]=randint(1,6)
print(thrownDiceNum)
print("Laskettu summa on",sum(thrownDiceNum))
print("Suurin numero oli",max(thrownDiceNum))
```
